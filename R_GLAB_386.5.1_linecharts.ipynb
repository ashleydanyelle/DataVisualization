```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.arange(1, 12, 2)
y = np.arange(2, 13, 2)

# create a line plot with a dashdot linestyle
plt.plot(x, y, linestyle='dashdot')
plt.show()
```

```python
plt.plot(x, y, linestyle=':', color='pink')
plt.show()
```

```python
x = np.array([3, 8, 1, 10, 4, 200, 45, 67, 56, 98, 100])
y = np.array([3, 8, 1, 10, 4, 200, 45, 67, 56, 98, 100])
```

```python
# Create a dashed line, with diamond markers
plt.plot(y, linestyle='--', marker='D', color='orange')

plt.show()
```

```python
# Using the opposite approach, linestyle of dashdot, and square markers
fig, ax = plt.subplots()

ax.plot(y, linestyle='dashdot', marker='s', color='purple')
```

```python
# Diamond marker, change the marker color, change marker width
plt.plot(y, marker='D', markerfacecolor='orange', markersize=10)
plt.show()
```

```python
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 170, 190, 160, 300, 310, 320, 330])
```

```python
# Create a lineplot, custome color, dotted linestyle, diamond markers
plt.plot(x, y, linestyle=':', marker='D', markersize=10, color='purple')
plt.title('Sports Watch Data')
plt.xlabel('Average Pulse')
plt.ylabel('Calorie Burnage')
plt.show()
```

```python
# Use hex codes
# Create a lineplot, custome color, dotted linestyle, diamond markers
plt.plot(x, y, linestyle=':', marker='D', markersize=10, color='#00CED1')
plt.title('Sports Watch Data')
plt.xlabel('Average Pulse')
plt.ylabel('Calorie Burnage')
plt.show()
```

```python
years=[1970,1980,1985,1990,1995,2000,2005,2010,2015,2020,2025,2030,2035]
pops=[3.6,4.0,4.4,4.8,5.3,5.7,6.1,6.5,7.3,7.3,7.3,7.4,7.5]
```

```python
# Create title of World Population - Blue
# Labels growth by year, population in billions, green red

# Line plot using shorthand format: 'g--D' = green line, dashed, diamond markers
plt.plot(years, pops, 'g--D')

# Titles and Labels with custom colors
plt.title('World Population', color='blue')
plt.xlabel('Growth by Year', color='green')
plt.ylabel('Population in Billions', color='red')
```

```python
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 170, 190, 160, 300, 310, 320, 330])

# Create a lineplot, custome color, dotted linestyle, diamond markers
plt.plot(x, y, linestyle=':', marker='D', markersize=10, color='blue')
plt.title('Sports Watch Data')
plt.xlabel('Average Pulse')
plt.ylabel('Calorie Burnage')
plt.grid()
plt.show()
```
